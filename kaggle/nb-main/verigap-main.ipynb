# VeriGap — Coding Agents Build Software They Cannot Test

### A repository-scale benchmark for test-suite generation from specifications

Given only a program **specification**, can a coding agent write a test suite strong enough to tell a correct implementation from a subtly-broken one? For an **offline agent on a single accelerator**, its own tests are its only reviewer — so this is the capability to measure first.

**The verification gap (frontier agents):** valid tests (88–96% pass the reference) and correct code (86–93% of reference tests pass), yet suites kill **at most 55% of requirement-targeted mutants** — and handing over the real program doesn't help. The failure is **test design**, not spec clarity.

## The three tasks (pool into the VeriGap benchmark)

| Task | Metric | Notebook |
|---|---|---|
| **Validity (V)** | tests passing the reference | `verigap-validity` |
| **Mutation (M)** | mutants killed by valid tests (headline) | `verigap-mutation` |
| **Code-Gen (R)** | reference tests passing the agent's build | `verigap-codegen` |

Each is a standalone, self-contained `@kbench.task` notebook with `verigap-specs` + `verigap-data` attached. Use **Add Models** on each to populate the cross-model leaderboard, then pool all three under the **VeriGap** benchmark.

## How scoring works (container-free)

A suite is a directory with `eval/run.sh` that emits JUnit `results.xml` and invokes `/workspace/executable`. Reference and mutant binaries are precompiled; Kaggle's `linux/amd64` runtime runs them **natively, no Docker**. `V` = tests passing the reference; a mutant is **killed** when a reference-passing test fails on it, `M` = killed/total; `R` = reference tests passing the agent's build.
